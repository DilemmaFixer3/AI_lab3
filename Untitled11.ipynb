```python
# Необхідні інсталяції, якщо не встановлено:
# pip install torch numpy scipy

import torch
import torch.nn as nn
import numpy as np
from torch.autograd import grad
from scipy.stats import uniform # Для LHS, якщо pyDOE недоступний

# Фіксація seed для відтворюваності
torch.manual_seed(1234)
np.random.seed(1234)

# Параметри задачі
DOMAIN_X = [0.0, 1.0]
DOMAIN_T = [0.0, 1.0]
ALPHA = 1.0 # Коефіцієнт температуропровідності

# Початкова та Граничні умови
def u_initial(x):
    # u(x, 0) = 2x(x+1) + 0.5
    return 2.0 * x * (x + 1.0) + 0.5

def u_boundary_0(t):
    # u(0, t) = 2t + 0.5
    return 2.0 * t + 0.5

def u_boundary_1(t):
    # u(1, t) = 2t + 4.5
    return 2.0 * t + 4.5
```

```python
def lhs_samples(n_dim, n_samples):
    """
    Проста реалізація Latin Hypercube Sampling (LHS) з використанням numpy та scipy.stats.
    """
    # Генеруємо зразки для кожного виміру
    samples = np.zeros((n_samples, n_dim))

    # Створюємо сітку
    for j in range(n_dim):
        samples[:, j] = np.random.permutation(n_samples) # Перестановка номерів страт

    # Масштабуємо та додаємо випадкові зсуви
    samples = (samples + uniform.rvs(size=(n_samples, n_dim))) / n_samples

    return samples
```

```python
class MLP(nn.Module):
    def __init__(self, layers):
        super(MLP, self).__init__()
        self.layers = nn.ModuleList()
        in_dim = layers[0]
        for out_dim in layers[1:]:
            self.layers.append(nn.Linear(in_dim, out_dim))
            nn.init.xavier_uniform_(self.layers[-1].weight)
            nn.init.zeros_(self.layers[-1].bias)
            in_dim = out_dim

        self.layers[-1] = nn.Linear(layers[-2], layers[-1])
        nn.init.xavier_uniform_(self.layers[-1].weight)
        nn.init.zeros_(self.layers[-1].bias)

    def forward(self, x):
        for i, layer in enumerate(self.layers[:-1]):
            x = torch.tanh(layer(x)) # Активація tanh
        x = self.layers[-1](x)
        return x

layers = [2] + [128] * 6 + [1]
# Початкова ініціалізація моделі, яка буде скопійована для експериментів
initial_model = MLP(layers)

# Функція для обчислення залишку PDE (Residual, f_res)
def pde_residual(model, xt_f):
    x = xt_f[:, 0:1].clone().detach().requires_grad_(True)
    t = xt_f[:, 1:2].clone().detach().requires_grad_(True)

    u = model(torch.cat([x, t], dim=1))

    # Автоматичне диференціювання
    u_t = grad(u, t, torch.ones_like(u), create_graph=True, retain_graph=True)[0]
    u_x = grad(u, x, torch.ones_like(u), create_graph=True, retain_graph=True)[0]
    u_xx = grad(u_x, x, torch.ones_like(u_x), create_graph=True, retain_graph=True)[0]

    # Залишок PDE: f = ∂u/∂t - α * ∂²u/∂x²
    f_res = u_t - ALPHA * u_xx
    return f_res
```

```python
N_f = 10000
N_b = 500
N_0 = 500

# 3.1. Генерація точок всередині області Ω (f-точки) за допомогою LHS
xt_f_np = lhs_samples(2, N_f)
xt_f_np[:, 0] = xt_f_np[:, 0] * (DOMAIN_X[1] - DOMAIN_X[0]) + DOMAIN_X[0]
xt_f_np[:, 1] = xt_f_np[:, 1] * (DOMAIN_T[1] - DOMAIN_T[0]) + DOMAIN_T[0]
xt_f = torch.tensor(xt_f_np, dtype=torch.float32, requires_grad=True)

# 3.2. Генерація точок для граничних умов (b-точки)
t_b_0 = lhs_samples(1, N_b) * (DOMAIN_T[1] - DOMAIN_T[0]) + DOMAIN_T[0]
x_b_0 = np.zeros_like(t_b_0)
xt_b_0 = np.hstack((x_b_0, t_b_0))
u_b_0_exact = u_boundary_0(torch.tensor(t_b_0, dtype=torch.float32))

t_b_1 = lhs_samples(1, N_b) * (DOMAIN_T[1] - DOMAIN_T[0]) + DOMAIN_T[0]
x_b_1 = np.ones_like(t_b_1)
xt_b_1 = np.hstack((x_b_1, t_b_1))
u_b_1_exact = u_boundary_1(torch.tensor(t_b_1, dtype=torch.float32))

xt_b = np.vstack((xt_b_0, xt_b_1))
xt_b = torch.tensor(xt_b, dtype=torch.float32, requires_grad=True)
u_b_exact = torch.cat((u_b_0_exact, u_b_1_exact))

# 3.3. Генерація точок для початкової умови (0-точки)
x_0 = lhs_samples(1, N_0) * (DOMAIN_X[1] - DOMAIN_X[0]) + DOMAIN_X[0]
t_0 = np.zeros_like(x_0)
xt_0 = np.hstack((x_0, t_0))
u_0_exact = u_initial(torch.tensor(x_0, dtype=torch.float32))
xt_0 = torch.tensor(xt_0, dtype=torch.float32, requires_grad=True)
```

```python
LAMBDA_F = 1.0
LAMBDA_0 = 1.0

def loss_func(model, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b):
    mse_loss = nn.MSELoss()

    # L_f (Residual Loss)
    f_pred = pde_residual(model, xt_f)
    loss_f = mse_loss(f_pred, torch.zeros_like(f_pred))

    # L_b (Boundary Loss)
    u_b_pred = model(xt_b)
    loss_b = mse_loss(u_b_pred.squeeze(), u_b_exact.squeeze())

    # L_0 (Initial Loss)
    u_0_pred = model(xt_0)
    loss_0 = mse_loss(u_0_pred.squeeze(), u_0_exact.squeeze())

    # L = λ_f * L_f + λ_b * L_b + λ_0 * L_0
    total_loss = LAMBDA_F * loss_f + lambda_b * loss_b + LAMBDA_0 * loss_0

    return total_loss, loss_f.item(), loss_b.item(), loss_0.item()

def train_model(initial_model_state, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b, optimizer_mode):
    model_copy = MLP(layers)
    model_copy.load_state_dict(initial_model_state)

    # ПРИСКОРЕННЯ: Зменшуємо кількість кроків Adam
    N_ADAM_EPOCHS_FAST = 3000   # Зменшено з 10000 для режиму Adam
    N_ADAM_WARMUP_FAST = 300    # Зменшено з 1000 для режиму Adam+L-BFGS
    N_LBFGS_ITER_FAST = 200     # Зменшено з 500 для режиму Adam+L-BFGS

    if optimizer_mode == 'Adam':
        optimizer = torch.optim.Adam(model_copy.parameters(), lr=1e-3)
        n_epochs = N_ADAM_EPOCHS_FAST # Використовуємо прискорену кількість епох

        for epoch in range(n_epochs):
            optimizer.zero_grad()
            total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
            total_loss.backward()
            optimizer.step()

            if (epoch + 1) % 1000 == 0:
                # Додатковий вивід для моніторингу, що не зависло
                print(f'Adam (λ_b={lambda_b}): Epoch {epoch+1}/{n_epochs}, Loss: {total_loss.item():.4e}, L_b: {lb:.4e}')

        return model_copy, total_loss.item(), {'L_f': lf, 'L_b': lb, 'L_0': l0}

    elif optimizer_mode == 'Adam+L-BFGS':
        # 1. Етап Adam (Warup)
        optimizer_adam = torch.optim.Adam(model_copy.parameters(), lr=1e-3)
        n_adam_epochs = N_ADAM_WARMUP_FAST # Короткий прогрів
        for _ in range(n_adam_epochs):
            optimizer_adam.zero_grad()
            total_loss, _, _, _ = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
            total_loss.backward()
            optimizer_adam.step()

        # 2. Етап L-BFGS (Зменшена кількість ітерацій)
        optimizer_lbfgs = torch.optim.LBFGS(model_copy.parameters(),
                                            max_iter=N_LBFGS_ITER_FAST, # Прискорення тут!
                                            line_search_fn='strong_wolfe')

        def closure():
            optimizer_lbfgs.zero_grad()
            total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
            total_loss.backward()
            # Цей вивід показує, що L-BFGS працює, незважаючи на "зависання"
            print(f'L-BFGS (λ_b={lambda_b}): Current Loss: {total_loss.item():.4e}, L_b: {lb:.4e}')
            return total_loss

        optimizer_lbfgs.step(closure)

        total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)

        return model_copy, total_loss.item(), {'L_f': lf, 'L_b': lb, 'L_0': l0}

    return model_copy, 0.0, {} # Заглушка на випадок помилки
```

```python
lambda_b_base = 10.0
# Скидаємо початковий стан для чесного порівняння, як в оригіналі
initial_state = initial_model.state_dict()

print("\n--- 1. Порівняння Adam проти Adam+L-BFGS (λ_b=10) [ПРИСКОРЕНЕ] ---")

# Тренування 1: Тільки Adam (зменшена кількість епох)
model_adam, loss_adam, components_adam = train_model(initial_state, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b_base, 'Adam')

# Тренування 2: Adam + L-BFGS (зменшений прогрів Adam та ітерації L-BFGS)
model_adam_lbfgs, loss_lbfgs, components_lbfgs = train_model(initial_state, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b_base, 'Adam+L-BFGS')

print("\n--- Результати Порівняння (λ_b=10) ---")
print(f"Adam: Фінальна Загальна Втрата: {loss_adam:.4e} | Втрата ГУ (L_b): {components_adam['L_b']:.4e}")
print(f"Adam+L-BFGS: Фінальна Загальна Втрата: {loss_lbfgs:.4e} | Втрата ГУ (L_b): {components_lbfgs['L_b']:.4e}")
```

```python
lambda_b_values = [5.0, 10.0, 20.0]
results_lambda_b = {}
best_model_for_flux = None

print("\n--- 2. Дослідження впливу λ_b на порушення ГУ (L2 Error BC) ---")

# Створимо тестову сітку ГУ
t_test = np.linspace(DOMAIN_T[0], DOMAIN_T[1], 100)
xt_b0_test = torch.tensor(np.array([[0.0, t] for t in t_test]), dtype=torch.float32)
xt_b1_test = torch.tensor(np.array([[1.0, t] for t in t_test]), dtype=torch.float32)
u_b0_exact_test = u_boundary_0(torch.tensor(t_test, dtype=torch.float32)).numpy()
u_b1_exact_test = u_boundary_1(torch.tensor(t_test, dtype=torch.float32)).numpy()

for lambda_b in lambda_b_values:
    # Тренування з Adam+L-BFGS (кращий оптимізатор)
    model_lambda, loss_total, components = train_model(initial_state, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b, 'Adam+L-BFGS')

    # Перевірка виконання ГУ (L2-похибка)
    u_b0_pred = model_lambda(xt_b0_test).squeeze().detach().numpy()
    u_b1_pred = model_lambda(xt_b1_test).squeeze().detach().numpy()

    l2_error_b0 = np.sqrt(np.mean((u_b0_pred - u_b0_exact_test)**2))
    l2_error_b1 = np.sqrt(np.mean((u_b1_pred - u_b1_exact_test)**2))
    l2_error_b = (l2_error_b0 + l2_error_b1) / 2

    results_lambda_b[lambda_b] = {'L_b_Loss': components['L_b'], 'L2_Error_BC': l2_error_b}
    print(f"λ_b={lambda_b}: L_b Loss: {components['L_b']:.4e} | Середня L2 Error BC: {l2_error_b:.4e}")

    if lambda_b == 20.0:
        best_model_for_flux = model_lambda # Вибираємо модель з найбільшою λ_b
```

```python
time_points = [0.1, 0.5, 0.9] # 3 моменти часу
best_model = best_model_for_flux

print("\n--- 3. Обчислення Теплового Потоку на Межах (модель λ_b=20) ---")

def calculate_heat_flux(model, x_val, t_val):
    # Вхідні змінні повинні мати requires_grad=True
    x = torch.tensor([[x_val]], dtype=torch.float32, requires_grad=True)
    t = torch.tensor([[t_val]], dtype=torch.float32, requires_grad=True)
    xt = torch.cat([x, t], dim=1)

    u = model(xt)

    # ∂u/∂x
    u_x = grad(u, x, torch.ones_like(u), create_graph=False)[0]

    # Тепловий потік q = - ∂u/∂x
    q = -u_x.item()
    return q

for t in time_points:
    q_0 = calculate_heat_flux(best_model, 0.0, t)
    q_1 = calculate_heat_flux(best_model, 1.0, t)

    print(f"Час t={t}:")
    print(f"  Потік на x=0 (вхід): q(0, {t}) = {q_0:.4e}")
    print(f"  Потік на x=1 (вихід): q(1, {t}) = {q_1:.4e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm # Для гарного прогрес-бару

# --- 1.1. Клас MLP з Sin (SIREN-like) активацією ---
class MLP_Sin(nn.Module):
    def __init__(self, layers, w0=30.0): # w0=30 для SIREN
        super(MLP_Sin, self).__init__()
        self.layers = nn.ModuleList()
        self.w0 = w0
        in_dim = layers[0]

        for out_dim in layers[1:]:
            lin = nn.Linear(in_dim, out_dim)

            # SIREN Ініціалізація
            if in_dim == 2: # Перший шар (x, t)
                nn.init.uniform_(lin.weight, -1 / in_dim, 1 / in_dim)
            else:
                nn.init.uniform_(lin.weight, -np.sqrt(6 / in_dim) / self.w0, np.sqrt(6 / in_dim) / self.w0)

            nn.init.zeros_(lin.bias)
            self.layers.append(lin)
            in_dim = out_dim

        self.layers[-1] = nn.Linear(layers[-2], layers[-1])
        nn.init.xavier_uniform_(self.layers[-1].weight)
        nn.init.zeros_(self.layers[-1].bias)

    def forward(self, x):
        for i, layer in enumerate(self.layers[:-1]):
            # Перший шар має w0=1, наступні w0=30
            current_w0 = 1.0 if i == 0 else self.w0
            x = torch.sin(current_w0 * layer(x))
        x = self.layers[-1](x)
        return x


# --- 1.2. Функція оцінювання (L2-похибка, RMSE) ---
def evaluate_metrics(model, xt_test, u_exact_test):
    """Обчислює L2-похибку та RMSE на тестовій сітці."""
    u_pred = model(xt_test).detach().numpy().flatten()
    u_exact = u_exact_test.numpy().flatten()

    # RMSE (Root Mean Square Error)
    rmse = np.sqrt(np.mean((u_pred - u_exact)**2))

    # Відносна L2-похибка (Relative L2 Error)
    # ||u_pred - u_exact||_2 / ||u_exact||_2
    relative_l2 = np.linalg.norm(u_pred - u_exact) / np.linalg.norm(u_exact)

    # Карта похибок
    error_map = np.abs(u_pred - u_exact)

    return {'RMSE': rmse, 'Relative_L2': relative_l2, 'Error_Map': error_map}

# --- 1.3. Модифікована функція тренування для збору кривих збіжності ---
def train_model_with_logging(initial_model_state, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b, optimizer_mode, model_class):
    model_copy = model_class(layers)
    model_copy.load_state_dict(initial_model_state)

    # Логування втрат
    loss_log = []

    N_ADAM_EPOCHS_FAST = 3000
    N_ADAM_WARMUP_FAST = 300
    N_LBFGS_ITER_FAST = 200

    if optimizer_mode == 'Adam':
        optimizer = torch.optim.Adam(model_copy.parameters(), lr=1e-3)
        n_epochs = N_ADAM_EPOCHS_FAST

        for epoch in tqdm(range(n_epochs), desc=f"Adam (λ={lambda_b})"):
            optimizer.zero_grad()
            total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
            total_loss.backward()
            optimizer.step()
            loss_log.append([total_loss.item(), lf, lb, l0])

        final_loss = total_loss.item()
        components = {'L_f': lf, 'L_b': lb, 'L_0': l0}

    elif optimizer_mode == 'Adam+L-BFGS':
        # 1. Етап Adam (Warup)
        optimizer_adam = torch.optim.Adam(model_copy.parameters(), lr=1e-3)
        n_adam_epochs = N_ADAM_WARMUP_FAST
        for _ in tqdm(range(n_adam_epochs), desc=f"Adam Warmup (λ={lambda_b})"):
            optimizer_adam.zero_grad()
            total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
            total_loss.backward()
            optimizer_adam.step()
            loss_log.append([total_loss.item(), lf, lb, l0])

        # 2. Етап L-BFGS
        optimizer_lbfgs = torch.optim.LBFGS(model_copy.parameters(), max_iter=N_LBFGS_ITER_FAST, line_search_fn='strong_wolfe')

        def closure():
            optimizer_lbfgs.zero_grad()
            total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
            total_loss.backward()
            loss_log.append([total_loss.item(), lf, lb, l0])
            return total_loss

        print(f"Starting L-BFGS (max_iter={N_LBFGS_ITER_FAST})...")
        optimizer_lbfgs.step(closure)

        total_loss, lf, lb, l0 = loss_func(model_copy, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b)
        final_loss = total_loss.item()
        components = {'L_f': lf, 'L_b': lb, 'L_0': l0}

    return model_copy, final_loss, components, pd.DataFrame(loss_log, columns=['Total_Loss', 'L_f', 'L_b', 'L_0'])
```

```python
# --- 2.1. Формування Тестової Сітки для Візуалізації та Оцінювання ---
N_TEST = 100 # Кількість точок по X
M_TEST = 100 # Кількість точок по T
x_test = np.linspace(DOMAIN_X[0], DOMAIN_X[1], N_TEST)
t_test_full = np.linspace(DOMAIN_T[0], DOMAIN_T[1], M_TEST)
X_test, T_test = np.meshgrid(x_test, t_test_full)
xt_test_flat = torch.tensor(np.hstack((X_test.flatten()[:, None], T_test.flatten()[:, None])), dtype=torch.float32)

# --- УВАГА: Для отримання ЕТАЛОННОГО розв'язку (u_exact_test)
# Цей код ПРИПУСКАЄ, що аналітичний розв'язок u_exact(x, t) вам відомий
# АБО ви використовуєте модель, навчену на FDM/FEM (що вимагає окремої реалізації)
# Для цілей демонстрації, тут використовується заглушка:
#
# Якщо аналітичний розв'язок невідомий, метрики L2/RMSE можуть бути обчислені
# лише як порівняння між різними PINN-моделями (що менш інформативно).
#
# Припустимо, що u_exact_test = модель_PINN_з_найменшою_втратою (на щільній сітці)
# для відносних метрик. Або використовуємо відомий аналітичний розв'язок,
# якщо б рівняння було простішим.
#
# !!! Оскільки аналітичний розв'язок для u_t = u_xx з такими ГУ/ПУ не є очевидним,
# для оцінки метрик нам потрібен еталон.
# Замість еталону, ми просто виводимо карту поля і похибки.

# --- 2.2. Запуск експериментів ---
lambda_b_opt = 20.0
initial_state = initial_model.state_dict() # Використовуємо ініціалізацію Tanh

results_activation = {}

# Експеримент 1: Tanh
print("\n### Експеримент 1: Активація Tanh (Adam+L-BFGS, λ_b=20.0) ###")
model_tanh, loss_tanh, comp_tanh, log_tanh = train_model_with_logging(
    initial_state, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b_opt, 'Adam+L-BFGS', MLP
)
results_activation['Tanh'] = {'Loss': loss_tanh, 'Components': comp_tanh, 'Log': log_tanh, 'Model': model_tanh}

# Експеримент 2: Sin (SIREN-like)
# Потрібна нова ініціалізація, оскільки ініціалізація SIREN відрізняється
model_sin_init = MLP_Sin(layers)
initial_state_sin = model_sin_init.state_dict()

print("\n### Експеримент 2: Активація Sin (Adam+L-BFGS, λ_b=20.0) ###")
model_sin, loss_sin, comp_sin, log_sin = train_model_with_logging(
    initial_state_sin, xt_f, xt_b, u_b_exact, xt_0, u_0_exact, lambda_b_opt, 'Adam+L-BFGS', MLP_Sin
)
results_activation['Sin'] = {'Loss': loss_sin, 'Components': comp_sin, 'Log': log_sin, 'Model': model_sin}


# --- 2.3. Збір даних для фінальної таблиці та візуалізації ---

# Вибираємо кращу модель для фінальних візуалізацій
best_model_final = model_tanh if loss_tanh < loss_sin else model_sin
best_activation = 'Tanh' if loss_tanh < loss_sin else 'Sin'

print(f"\n--- Фінальний вибір: {best_activation} з втратою {min(loss_tanh, loss_sin):.4e} ---")

# Обчислення поля U(x, t)
u_pred_flat = best_model_final(xt_test_flat).detach().numpy().flatten()
u_pred_map = u_pred_flat.reshape(M_TEST, N_TEST)

# Для профілів u(x, t_k)
time_slices = [0.1, 0.5, 0.9] # 3 моменти часу для профілів
profiles = {}

for t_slice in time_slices:
    # Знаходимо найближчий індекс часу
    t_idx = np.argmin(np.abs(t_test_full - t_slice))

    profiles[f't={t_slice:.1f}'] = {
        'x': x_test,
        'u_pred': u_pred_map[t_idx, :],
        # 'u_exact': u_exact_map[t_idx, :]. Якщо відомий еталон
    }

# --- 2.4. Обчислення теплових потоків на межах (повторне обчислення, якщо потрібно) ---
# Використовуємо найкращу модель
print("\n--- Фінальний розрахунок теплових потоків (з кращою моделлю) ---")
flux_results_final = []
for t in time_points:
    q_0 = calculate_heat_flux(best_model_final, 0.0, t)
    q_1 = calculate_heat_flux(best_model_final, 1.0, t)
    flux_results_final.append({'t': t, 'q(0, t)': q_0, 'q(1, t)': q_1})
    print(f"Час t={t}: q(0, t) = {q_0:.4e}, q(1, t) = {q_1:.4e}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Встановлення стилю Matplotlib
plt.style.use('seaborn-v0_8-whitegrid')

# Припускаємо, що ці дані вже згенеровані попереднім кодом:
# u_pred_map, X_test, T_test, profiles, flux_results_final, results_activation
# Також потрібен u_exact_map для карти похибок.
# Оскільки еталонний розв'язок невідомий, ми створимо фіктивну карту похибок
# для демонстрації або використовуватимемо лише карту поля U(x,t).

# --- А. Карта Поля u(x, t) ---

def plot_solution_map(X, T, U_map, title="Карта поля температури u(x,t)"):
    """Візуалізація поля температури у двомірному просторі (x, t)."""
    fig, ax = plt.subplots(1, 1, figsize=(8, 6))

    # Використовуємо pcolormesh для кращої візуалізації сітки
    h = ax.pcolormesh(T, X, U_map.T, cmap='viridis', shading='auto')

    fig.colorbar(h, ax=ax, label='Температура, u')
    ax.set_xlabel('Час, t')
    ax.set_ylabel('Простір, x')
    ax.set_title(title)
    plt.show()

print("\n--- Генерація Карти Поля u(x, t) ---")
plot_solution_map(X_test, T_test, u_pred_map,
                  title=f"Карта поля u(x,t) (Активація: {best_activation})")


# --- Б. Карта Похибок (Error Map) ---

# !!! Важливо: Якщо еталонний розв'язок (u_exact_map) не відомий,
# ця карта не може бути побудована. Використовуємо фіктивний "еталон"
# для демонстрації, якби він був.

# Створимо фіктивний масив похибки для візуалізації
# error_map_fake = np.abs(u_pred_map - np.roll(u_pred_map, 10, axis=1))
# plot_solution_map(X_test, T_test, error_map_fake,
#                   title="Фіктивна Карта Похибок |u_pred - u_exact|",
#                   cmap='Reds')

# Використовуємо реальну карту похибки з results_activation, якщо вона є
# Якщо ви додали еталонний розв'язок, використайте його:
# error_map = np.abs(u_pred_map - U_EXACT_MAP)
# plot_solution_map(X_test, T_test, error_map, title="Карта Похибок", cmap='Reds')


# --- В. Профілі u(x, tk) для 3 моментів часу ---

def plot_profiles(profiles_data):
    """Візуалізація профілів температури у просторі для різних моментів часу."""
    fig, ax = plt.subplots(figsize=(8, 6))

    for t_label, data in profiles_data.items():
        # u_exact (якщо є): ax.plot(data['x'], data['u_exact'], '--', label=f'Еталон ({t_label})')
        ax.plot(data['x'], data['u_pred'], label=f'PINN ({t_label})', linewidth=2)

    ax.set_xlabel('Просторова координата, x')
    ax.set_ylabel('Температура, u')
    ax.set_title('Профілі температури u(x, t) у часі')
    ax.legend()
    plt.show()

print("\n--- Генерація Профілів u(x, tk) ---")
plot_profiles(profiles)


# --- Г. Криві Збіжності Складників Втрат ---

def plot_convergence_curves(log_data, title_suffix):
    """Візуалізація еволюції втрат у часі."""
    df = log_data.reset_index().rename(columns={'index': 'Epoch'})

    fig, ax = plt.subplots(figsize=(10, 6))

    ax.plot(df['Epoch'], df['Total_Loss'], label='Total Loss (Загальна Втрата)', color='black', linewidth=2)
    ax.plot(df['Epoch'], df['L_f'], label='PDE Loss (L_f)', linestyle='--', alpha=0.7)
    ax.plot(df['Epoch'], df['L_b'], label='Boundary Loss (L_b)', linestyle='--', alpha=0.7)
    ax.plot(df['Epoch'], df['L_0'], label='Initial Loss (L_0)', linestyle='--', alpha=0.7)

    ax.set_yscale('log') # Втрати зазвичай показують у логарифмічній шкалі
    ax.set_xlabel('Епоха / Крок (Adam) / Ітерація (L-BFGS)')
    ax.set_ylabel('Втрата (логарифмічна шкала)')
    ax.set_title(f'Криві збіжності PINN ({title_suffix})')
    ax.legend()
    plt.show()

print("\n--- Генерація Кривих Збіжності (Tanh vs Sin) ---")
plot_convergence_curves(results_activation['Tanh']['Log'], "Tanh активація")
plot_convergence_curves(results_activation['Sin']['Log'], "Sin активація (SIREN-like)")


# --- Д. Теплові Потоки на Межах ---

def plot_heat_flux(flux_data):
    """Візуалізація теплових потоків на межах."""
    df = pd.DataFrame(flux_data)

    fig, ax = plt.subplots(figsize=(8, 6))

    ax.plot(df['t'], df['q(0, t)'], 'o-', label='Потік на x=0 ($q_0$)', linewidth=2)
    ax.plot(df['t'], df['q(1, t)'], 's-', label='Потік на x=1 ($q_1$)', linewidth=2)

    ax.axhline(0, color='gray', linestyle='--', linewidth=0.5)

    ax.set_xlabel('Час, t')
    ax.set_ylabel('Тепловий потік, q = -∂u/∂x')
    ax.set_title('Тепловий потік на граничних межах')
    ax.legend()
    plt.show()

print("\n--- Генерація Графіка Теплового Потоку ---")
plot_heat_flux(flux_results_final)
```
